```python
import numpy as np

a = np.array([4,7,2,1,5])

print("Sum =", np.sum(a))
print("Mean =", np.mean(a))
print("Maximum =", np.max(a))
print("Square Root =", np.sqrt(a))
print("Sorted =", np.sort(a))
```

```python
import numpy as np
x = np.pi / 4
print("Sine:", np.sin(x))
print("Cosine:", np.cos(x))
print("Tangent:", np.tan(x))
print("Inverse Sine:", np.arcsin(0.707))
print("Inverse Cosine:", np.arccos(0.707))
print("Inverse Tangent:", np.arctan(1))
print("Radians to Degrees:", np.degrees(x))
print("Degrees to Radians:", np.radians(45))
```
